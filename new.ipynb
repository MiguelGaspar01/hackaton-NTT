```python
month_map = {
    "jan": 1, "feb": 2, "mar": 3, "apr": 4,
    "may": 5, "jun": 6, "jul": 7, "aug": 8,
    "sep": 9, "oct": 10, "nov": 11, "dec": 12
}

X_train["month"] = X_train["month"].map(month_map)
X_test["month"] = X_test["month"].map(month_map)

def cyclical_encode(x, max_val):
    return np.column_stack([
        np.sin(2 * np.pi * x / max_val),
        np.cos(2 * np.pi * x / max_val)
    ])

month_encoder = FunctionTransformer(lambda x: cyclical_encode(x, 12))
day_encoder   = FunctionTransformer(lambda x: cyclical_encode(x, 31))

preprocess = ColumnTransformer(
    transformers=[
        ("month_cyc", month_encoder, ["month"]),
        ("day_cyc", day_encoder, ["day"])
    ],
    remainder="drop"
)

preprocess.fit(X_train)
X_train = preprocess.transform(X_train)
X_test = preprocess.transform(X_test)
#X.drop(columns=["month", "day"], inplace=True)
```
